### 2.3.2.1 Agriculture Dataset – Numerical Validation

```python
# Check numerical value ranges

agriculture_checks = {
    'soil_ph': (0, 14),
    'temperature_c': (-10, 60),
    'humidity_pct': (0, 100),
    'rainfall_mm': (0, None),
    'nitrogen_kg_ha': (0, None),
    'phosphorus_kg_ha': (0, None),
    'potassium_kg_ha': (0, None),
    'yield_tons_ha': (0, None)
}

for column, (lower, upper) in agriculture_checks.items():

    invalid = pd.Series(False, index=df_agriculture.index)

    if lower is not None:
        invalid |= df_agriculture[column] < lower

    if upper is not None:
        invalid |= df_agriculture[column] > upper

    print(
        column,
        "| Invalid records:",
        invalid.sum()
    )
```

```python

import pandas as pd

required_states = [
    "Maharashtra",
    "Telangana",
    "Karnataka"
]

required_crops = [
    "Rice",
    "Soybean",
    "Cotton",
    "Sugarcane",
    "Tomato"
]

# Filter required states and crops
filtered = agriculture_clean[
    agriculture_clean["state"].isin(required_states)
    & agriculture_clean["crop"].isin(required_crops)
]

# Create coverage table
coverage = pd.crosstab(
    filtered["state"],
    filtered["crop"]
)

coverage = coverage.reindex(
    index=required_states,
    columns=required_crops,
    fill_value=0
)

print("STATE-WISE CROP COVERAGE")
display(coverage)

print("\nMISSING COMBINATIONS")
for state in required_states:
    for crop in required_crops:
        if coverage.loc[state, crop] == 0:
            print(f"{state} - {crop}: Missing")
```

```python

import pandas as pd
import matplotlib.pyplot as plt

# Five crops required by the client
required_crops = [
    "Rice", "Soybean", "Cotton",
    "Sugarcane", "Tomato"
]

# Select relevant crop records
crop_data = agriculture_clean[
    agriculture_clean["crop"].isin(required_crops)
].copy()

# Calculate rainfall-yield correlation per crop
correlations = []

for crop in required_crops:
    subset = crop_data[crop_data["crop"] == crop]

    corr = subset["rainfall_mm"].corr(
        subset["yield_tonnes_per_ha"]
    )

    correlations.append({
        "Crop": crop,
        "Records": len(subset),
        "Rainfall_Yield_Correlation": corr
    })

correlation_df = pd.DataFrame(correlations)

print("RAINFALL VS CROP YIELD CORRELATION")
display(correlation_df.round(3))

# Plot correlation values
plt.figure(figsize=(9, 5))

plt.bar(
    correlation_df["Crop"],
    correlation_df["Rainfall_Yield_Correlation"]
)

plt.axhline(0, color="black", linewidth=0.8)
plt.title("Rainfall vs Crop Yield Correlation")
plt.xlabel("Crop")
plt.ylabel("Pearson Correlation")
plt.ylim(-1, 1)
plt.tight_layout()
plt.show()
```

```python

import matplotlib.pyplot as plt
import seaborn as sns

required_crops = [
    "Rice", "Soybean", "Cotton",
    "Sugarcane", "Tomato"
]

fig, axes = plt.subplots(2, 3, figsize=(15, 9))
axes = axes.flatten()

for i, crop in enumerate(required_crops):
    subset = crop_data[crop_data["crop"] == crop]

    sns.regplot(
        data=subset,
        x="rainfall_mm",
        y="yield_tonnes_per_ha",
        ax=axes[i],
        scatter_kws={"alpha": 0.6},
        line_kws={"color": "red"}
    )

    corr = subset["rainfall_mm"].corr(
        subset["yield_tonnes_per_ha"]
    )

    axes[i].set_title(f"{crop} (r = {corr:.3f})")
    axes[i].set_xlabel("Rainfall (mm)")
    axes[i].set_ylabel("Yield (tonnes/ha)")

axes[5].axis("off")

plt.suptitle(
    "Rainfall vs Crop Yield - Five Major Crops",
    fontsize=16
)

plt.tight_layout()
plt.show()
```

```python

import pandas as pd
from pathlib import Path
from io import StringIO
import re

raw_dir = Path.cwd().parent / "data" / "raw"
processed_dir = Path.cwd().parent / "data" / "processed"
processed_dir.mkdir(parents=True, exist_ok=True)

reports = {
    "Cereals & Millets": "apy_total_group_report _cereals_3states.xls",
    "Foodgrains": "apy_total_group_report_3foodgrains.xls",
    "Oilseeds": "apy_total_group_report_3oilseeds.xls",
    "Pulses": "apy_total_group_report_3pulses.xls"
}

all_data = []

for group, filename in reports.items():
    html = (raw_dir / filename).read_text(
        encoding="utf-8-sig",
        errors="replace"
    )

    df = pd.read_html(StringIO(html), flavor="lxml")[0]

    # Columns are a two-level header: year + measurement
    df.columns = [
        (str(a).strip(), str(b).strip())
        for a, b in df.columns
    ]

    state_col = next(c for c in df.columns if c[0] == "State")
    district_col = next(c for c in df.columns if c[0] == "District")

    years = sorted({
        c[0] for c in df.columns
        if re.fullmatch(r"\d{4}\s*-\s*\d{4}", c[0])
    })

    for year in years:
        subset = pd.DataFrame({
            "state": df[state_col],
            "district": df[district_col],
            "crop_group": group,
            "crop_year": year.replace(" ", ""),
            "area_hectares": df[(year, "Area (Hectare)")],
            "production_tonnes": df[(year, "Production (Tonnes)")],
            "yield_tonnes_per_hectare":
                df[(year, "Yield (Tonne/Hectare)")]
        })

        all_data.append(subset)

government_clean = pd.concat(all_data, ignore_index=True)

for col in [
    "area_hectares",
    "production_tonnes",
    "yield_tonnes_per_hectare"
]:
    government_clean[col] = pd.to_numeric(
        government_clean[col],
        errors="coerce"
    )

government_clean = government_clean.dropna(
    subset=["state", "district", "crop_year"]
)

government_clean = government_clean.drop_duplicates()

output = processed_dir / "government_crop_groups_clean.csv"
government_clean.to_csv(output, index=False)

print("Saved:", output)
print("Total rows:", len(government_clean))
print("\nStates:")
print(government_clean["state"].value_counts())
print("\nCrop groups:")
print(government_clean["crop_group"].value_counts())
print("\nMissing values:")
print(government_clean.isna().sum())
print("\nPreview:")
print(government_clean.head())
```

```python

import pandas as pd
import matplotlib.pyplot as plt
from pathlib import Path

project_dir = Path.cwd().parent
processed_dir = project_dir / "data" / "processed"
figures_dir = project_dir / "reports" / "figures"
figures_dir.mkdir(parents=True, exist_ok=True)

df = pd.read_csv(
    processed_dir / "government_crop_groups_clean.csv"
)

# Remove records with missing measurements
clean_df = df.dropna(
    subset=[
        "area_hectares",
        "production_tonnes",
        "yield_tonnes_per_hectare"
    ]
).copy()

# Remove invalid measurements
clean_df = clean_df[
    (clean_df["area_hectares"] > 0) &
    (clean_df["production_tonnes"] >= 0) &
    (clean_df["yield_tonnes_per_hectare"] >= 0)
].copy()

clean_df.to_csv(
    processed_dir / "government_crop_groups_eda.csv",
    index=False
)

print("Original rows:", len(df))
print("Clean rows:", len(clean_df))
print("Removed rows:", len(df) - len(clean_df))

# Weighted yield by state and crop group
summary = (
    clean_df.groupby(["state", "crop_group"])
    [["area_hectares", "production_tonnes"]]
    .sum()
    .reset_index()
)

summary["weighted_yield"] = (
    summary["production_tonnes"] /
    summary["area_hectares"]
)

summary.to_csv(
    processed_dir / "government_state_group_summary.csv",
    index=False
)

print("\nSTATE-WISE CROP GROUP SUMMARY")
print(summary.to_string(index=False))

# Chart 1: Crop-group yield comparison
pivot = summary.pivot(
    index="crop_group",
    columns="state",
    values="weighted_yield"
)

ax = pivot.plot(
    kind="bar",
    figsize=(11, 6),
    rot=0
)

ax.set_title("Government Crop Group Yield by State")
ax.set_xlabel("Crop Group")
ax.set_ylabel("Production / Area (Tonnes per Hectare)")
ax.legend(title="State")
plt.tight_layout()

plt.savefig(
    figures_dir / "government_crop_group_yield.png",
    dpi=300
)
plt.show()

# Chart 2: Yield trend across agricultural years
yearly = (
    clean_df.groupby(["crop_year", "crop_group"])
    [["area_hectares", "production_tonnes"]]
    .sum()
    .reset_index()
)

yearly["weighted_yield"] = (
    yearly["production_tonnes"] /
    yearly["area_hectares"]
)

year_pivot = yearly.pivot(
    index="crop_year",
    columns="crop_group",
    values="weighted_yield"
).sort_index()

ax = year_pivot.plot(
    marker="o",
    figsize=(11, 6)
)

ax.set_title("Crop Group Yield Trends (2020-2023)")
ax.set_xlabel("Agricultural Year")
ax.set_ylabel("Production / Area (Tonnes per Hectare)")
ax.grid(alpha=0.3)
plt.tight_layout()

plt.savefig(
    figures_dir / "government_yield_trends.png",
    dpi=300
)
plt.show()

print("\nEDA completed. Charts saved.")
```
